
Neste projeto, desenvolvo um modelo capaz de prever os níveis de ozônio nos próximos meses para uma empresa de consultoria ambiental. Utilizando a biblioteca Prophet, crio e valido o modelo e também construo uma aplicação web que permite ao usuário inserir um período de previsão e visualizar os resultados.

A aplicação gera um gráfico interativo com a série temporal e as previsões, além de uma tabela com os valores estimados, que pode ser baixada em formato CSV para análises futuras e tomada de decisões.

## Bibliotecas


```python
import pandas as pd
import plotly.express as px
import plotly.express as px
```

### Analise Exploratória


```python
df = pd.read_csv('https://raw.githubusercontent.com/alura-cursos/series_temporais_prophet/main/Dados/poluentes.csv')
df
```

```python
##tratamento data
df['Data'] = pd.to_datetime(df['Data'])
```

```python
###Padrao de datas sempre no começo do ano a camada de O3 está baixa

fig = px.line(df, x='Data', y='O3')
fig.show()
```

```python
##Media de mês a mês nos dados em 2022

df_2022 = df[df['Data'].dt.year == 2022]
df_2022_mensal = df_2022.groupby(df_2022['Data'].dt.month)['O3'].mean().reset_index()
import calendar
df_2022_mensal['Mês'] = df_2022_mensal['Data'].apply(lambda x: calendar.month_abbr[x])
fig = px.bar(df_2022_mensal, x='Mês', y='O3', title='Média de O3 ug/m3 por mês em 2022')
fig.show()
```
